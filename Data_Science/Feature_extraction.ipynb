## Principle Component Analysis (PCA)

Its an unsupervised machine learning problem<br>
Its an complex technique<br>
Underline Maths is too complex

Main goal is to reduce Curse of dimentionality

PCA tries to bring the higher dimention dataset to lower dimention<br>
The best thing about PCA is its takes the best possible lower dimention in consideration so that we can capture the essence of that higher dimention data into lower dimention <br>
So that when the ML model works on the lower dimention dataset the output will be good only

Benifits of using PCA
<br>
1. Faster execution of the Algorithm
2. Visualization

### Geometric Intuition

#### Feature Selection <br>
Selecting the imortant columns and removing the rest which will not help in prediction<br><br>

How this work mathametically<br>
Consider we have an scatter plot : here on there are multiple plots on the graph so ,the graph has an x axis and y axis  and on X axis there will me one of the columns and on Y there will be other <br>
So after plotting i will have to see that which one of the axis has the wider spread<br>
It means that consider X axis i will see the staring point and ending point on the X axis and measure its width <br>
Same way i will calculate the width of the starting point on the Y axis and its ending point <br>
Then we will compare both the width of X and width of Y Which is greater<br><br>
After comparing we will select that axis/column having more spread/width on graph<br><br>
This is how Feature Selection Works

#### Feature Extraction <br>

In some cases using Feature Selection does not work ,i.e we are not able to dicide which columns to consider and which to not.<br>
So in such case We use Feature Extraction<br><br>

Main idea is : we create an new columns which tells us about the number of columns we are about to merge into one <br><br>

In short this is what PCA does he forgots the existing features/columns and create a new set of featuers and from that new set of features PCA select the most important/best subset of features
<br><br>

How PCA is performed<br>
In the scatter graph , what PCA does is find new set of co-ordinate axis , it will rotate the axis in such a way that an new set of axis is formed<br>
So in this new set of Co-ordinate any one axis spread will change and then we can compre the width of the new co-ordinate axis and select the best feature/column <br><br>
When we rotate the original axis and get the new axis , we call the new axis as Principal Component 1 (PC1) and Principal Component 2 (PC2)<br>
After selecting the Principal component , we will transform that PC<br>
No.of Pc <= No.of original features

Every point in the scatter plot has an x axis and an y axis , we can consider points as vector which has 2 components, 1 in x direction and other in y direction,<br> 
so now we have to project this vector in any other vector , <br><br> so we have to find unit vector of the other vector , we represent unit vector as u <br>
so to find the point of the old vector on the new vector we use vector projection <br><br>
Formula to find vector projection is :  u*X/magnitude of u
<br><br> Since its unit vector so the magnitude of u will be 1 , so we can write the formula as<br><br>
formula : (u * x) or we can say (u transpose(u^t) * x)<br><br>
so the projection we will get after the formula will be an scalar value or we can say distance/length<br><br>

So what we have to do is maximize the Variance <br>
how we will calculate variance<br>

(1/n) Σ (xi-x')^2 

xi = u^t
x' = original data mean when pointer on the u it will become u^t*xmean or (u^tx')<br><br>

so now variance = (from i=1 to n)Σ ( u^txi - u^tx' )^2 / n <br><br>

Now using mathmetical we have to find such (u) for this the variance is maximum<br>
So we can say we will find the best u from all posibilities

## Co-variance and co-variance matrix <br>
Mean does not tell as how much the data is spread , so to solve this problem Variance came <br>
But variance has an problem too , the spread on both the axis on the graph can be difference , so variance does not show us the relationship between the two axis or feature.<br><br>

So here comes Co-variance, it shows the relation ship between both the axis <br>
formula = (x*y) for all points / no.of points
<br>
Its is closely related to co-relation<br>
only difference is we have range in co-relation of [-1,0,1] , in co-variance we dont have such thing<br><br>

So in short we can say that co-variance is used to find how axis behave regarding to each other / relationship of axis with each other.
<br><br><br>
Co-Variance Matrix<br>
Suppose we have 2D data,so it will be 2x2 matrix like
$$
\begin{bmatrix} cov(x1,x1) & cov(x1,x2) \\ cov(x2,x1) & cov(x2,x2) \end{bmatrix}
$$   

cov = Co-Variance <br>
cov(x1,x1) is var(x1) same for cov(x2,x2)<br>
cov(a,b) = cov(b,a)<br>
So its an square matrix ans Symmetric too<br>
Diagonal of this matrix are var(x) and other elements are cov(x,y)<br>
Adv of Cov  = It shows the spread of all the axis as it has variance of all the axis<br>
and other it will show the relationship of all the axis , its negative or positive<br><br>


#### Linear Transformation ,Eigen Vector and Eigen Values

How does Matrices work as Linear Transformation <br>
When we apply Matrices transformation , all the vector in the co-ordinate system , the direction and magnitude changes , basically they are getting thrown of their span (axis)<br>

So we can say that matrics are linear transformation which when applied to co-ordinate system then the vector in the co-ordinate system changes the magnitude and span. 

What are Eigen vector<br>
They are special vector ,which when transformation is applied ,there is not changes in them<br>
So when we applie Linear transformatino we get atleast 2 eigen vector,which have same direction only difference can be magnitude/scale<br>

A * V = lambda V<br>
where A = matrix<br>
V = eigen vector<br>
lambda = eigen value<br>
Its means that when we apply linear transforation on this vector then its as if that we are multiplying the vector with any scale/value<br>
This realtionship is always true for eigen vector

What are Eigen Values<br>
Eigen vector streach or shrink , the value by which they strach ot shrink is called Eigen values

Why we use Eigen vector and Eigen value in PCA<br><br>
the highest Eigen vector tell us the highest unit variance and that has the  highest variance 

### Step by Step Solution of PCA

we have an dataset with 3 columne F1,F2,F3 <br>
so in scatter plot its in 3D space and we have to bring it in lower dimention <br><br>

Step 1. Mean Center the data (Find the mean and minus with all data points)<br>
Step 2. Find co-variance matrix
<br>
Step 3. Find the Eigen value and Eigen vector for the co-variance matrix 
<br>highest Eigen vector value then PC1 and so on.
<br>So its upon us to either merge PC1 and PC2 and go in 2D or only select PC1 and do it in 1D.

### Coding Example

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(23)
mu_vec = np.array([0,0,0])
cov_mat1 = np.array([[1,0,0],[0,1,0],[0,0,1]])

class_sample = np.random.multivariate_normal(mu_vec,cov_mat1,20)

df = pd.DataFrame(class_sample,columns=['F1','F2','F3'])
df['target'] = 1

mu_vec2 = np.array([0,0,0])
cov_mat2 = np.array([[1,0,0],[0,1,0],[0,0,1]])

class_sample2 = np.random.multivariate_normal(mu_vec2,cov_mat2,20)

df2 = pd.DataFrame(class_sample2,columns=['F1','F2','F3'])
df2['target'] = 0

df = pd.concat((df,df2),ignore_index=True)
df = df.sample(40)
```

```python
df.head()
```

```python
import plotly.express as px

fig = px.scatter_3d(
    df,
    x=df['F1'],
    y=df['F2'],
    z=df['F3'],
    color=df['target'].astype(str)
)

fig.update_traces(
    marker=dict(
        size=12,
        line=dict(
            width=2,
            color='DarkSlateGrey'
        )
    )
)

fig.show()
```

```python
# step 1 Apply standard scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

df.iloc[:,0:3] = scaler.fit_transform(df.iloc[:,0:3])
```

```python
#step 2 finding Co-variance matrix
covariance_matrix = np.cov([df.iloc[:,0],df.iloc[:,1],df.iloc[:,2]])
covariance_matrix
```

```python
# step 3 finding eigne value and eigne vector
eigne_values,eigne_vector = np.linalg.eig(covariance_matrix)
eigne_values,eigne_vector
```

```python
pc = eigne_vector[0:2]
pc
```

```python
transformed_df = np.dot(df.iloc[:,0:3],pc.T)
new_df = pd.DataFrame(transformed_df,columns=['PC1','PC2'])
new_df['target'] = df['target'].values
new_df.head()
```

```python
new_df['PC1'] = new_df['PC1'].apply(lambda x: x.real)
new_df['PC2'] = new_df['PC2'].apply(lambda x: x.real)
```

```python
import plotly.express as px
new_df['target'] = new_df['target'].astype(str)

fig = px.scatter(
    x=new_df['PC1'],
    y=new_df['PC2'],
    color=new_df['target'].astype(str)
)

fig.update_traces(
    marker=dict(
        size=12,
        line=dict(
            width=2,
            color='DarkSlateGrey'
        )
    )
)

fig.show()
```

### Praticl example on MNIST Dataset

```python
df = pd.read_csv("train_image.csv")
df.head()
```

```python
X = df.iloc[:,1:]
Y = df.iloc[:,0]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train.shape
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier()
```

```python
knn.fit(X_train,y_train)
```

```python
import time 
start = time.time()
y_pred = knn.predict(X_test)
print(time.time() - start)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

#### Now we will do the PCA and repeat the same model algorithm

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
#PCA 
from sklearn.decomposition import PCA
pca = PCA(n_components=100) # how many best columns we need
```

```python
X_train_trf = pca.fit_transform(X_train)
X_test_trf = pca.transform(X_test)
```

```python
X_train_trf.shape
```

```python
knn = KNeighborsClassifier()
knn.fit(X_train_trf,y_train)
```

```python
y_pred = knn.predict(X_test_trf)
```

```python
accuracy_score(y_test,y_pred)
```

So we will noe change the component size and find the best component size or we can say u

```python
for i in range(1,785):
    pca = PCA(n_components=i)
    X_train_trf = pca.fit_transform(X_train)
    X_test_trf = pca.transform(X_test)
    knn = KNeighborsClassifier()
    knn.fit(X_train_trf,y_train)
    y_pred = knn.predict(X_test_trf)
    print(accuracy_score(y_test,y_pred))
    
```
