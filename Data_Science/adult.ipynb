```python
import pandas as pd
import numpy as np
```

```python
columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education-num",
    "marital-status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital-gain",
    "capital-loss",
    "hours-per-week",
    "native-country",
    "income"
]

df = pd.read_csv("adult.data", header=None, names=columns)

df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le= LabelEncoder()
```

```python
le.fit_transform(df["income"])
```

## Ordinal Encoding into the X columns 

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df["education"].unique()
```

```python
oe = OrdinalEncoder(categories=[ [   
    ' Preschool',
    ' 1st-4th',
    ' 5th-6th',
    ' 7th-8th',
    ' 9th',
    ' 10th',
    ' 11th',
    ' 12th',
    ' HS-grad',
    ' Some-college',
    ' Assoc-voc',
    ' Assoc-acdm',
    ' Bachelors',
    ' Masters',
    ' Prof-school',
    ' Doctorate']
    ])
```

```python
oe.fit_transform(df[["education"]])
```

```python
df["education"] = oe.fit_transform(df[["education"]])
df
```

## OneHot Encoder For Unorder Column

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(sparse_output=False,drop="first")
```

```python
encode=ohe.fit_transform(df[["workclass","marital-status","occupation","relationship","race","sex","native-country"]])
```

```python
encode_df=pd.DataFrame(encode,columns=ohe.get_feature_names_out(["workclass","marital-status","occupation","relationship","race","sex","native-country"]))
```

```python
encode_df.shape
```

```python
df= pd.concat([encode_df,df],axis=1)
```

```python
df
```

```python
df.drop(["workclass","marital-status","occupation","relationship","race","sex","native-country"],axis=1,inplace=True)
```
