Part B :  Missing Value

```python
missing_count=df.isnull().sum()
missing_count
```

```python
df[df.isna().any(axis=1)] # to find rows with missing values
```

```python
missing_sum=df.isna().sum().sum()
missing_sum # finding total missing number of cells
```

```python
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df.head()
```

```python
furniture_median = df.loc[df['Category'] == 'Furniture', 'Price'].median()
df['Price'] = df['Price'].fillna(furniture_median)
print("Furniture median price:", furniture_median)
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

```python
df[df.duplicated(subset=["Order_ID"])]
```

```python
#Missing Unique Identifier: A true duplicate check requires the Order_ID (or a combination of Order_ID + Timestamp). 
#Without checking the unique identifier, you cannot distinguish between a repeated event and a data entry error.
```

Part G :Management Question

```python
total_sales_by_city = df.groupby('City')['Sales'].sum()

best_city = total_sales_by_city.idxmax()
highest_sales = total_sales_by_city.max()
print(f"City with highest sales: {best_city}")
print(f"Total Sales: {highest_sales}")
```

```python
total_sales_by_city = df.groupby('Category')['Sales'].sum()

best_city = total_sales_by_city.idxmax()
highest_sales = total_sales_by_city.max()
print(f"Category with highest sales: {best_city}")
print(f"Total Sales: {highest_sales}")
```
