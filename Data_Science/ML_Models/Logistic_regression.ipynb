```python
import pandas as pd
import numpy as np
```

```python
from sklearn.model_selection import train_test_split
```

```python
class Logistic_regression:
    def __init__(self,alpha=0.01,n_init=1000):
        self.b = None
        self.m = None
        self.alpha = alpha
        self.n_init = n_init

    def _sigmoid(self,z):
        return (1 / (1+np.exp(-z)))

    def fit(self,X,y):
        r,c = X.shape

        self.b = 0
        self.m = np.zeros(c)

        for i in range(self.n_init):
            z = self.b + np.dot(X , self.m)
            y_pred = self._sigmoid(z)

            d_b = (1/r) * (y_pred - y)
            d_m = (1/r) * np.dot(X.T , (y_pred-y))

            self.b = self.b - self.alpha * d_b
            self.m = self.m - self.alpha * d_m
            
            

    def get_probabailities(self,x):
        # X = all the input columns , x = every single input in input column
        z = self.b + np.dot(x, self.m)
        prob_y = self._sigmoid(z)
        # this value is used to check if >=0.5 then 1 or else 0 class
        return prob_y

    
    def predict(self,X, threshold=0.5):
        probibalities = self.get_probabailities(X) 
        y_pred_bool = probibalities >= threshold
        return y_pred_bool.astype(int)
        
            
```

```python
X = np.array([[1,2],[2,3],[3,4],[4,5]])
y = np.array([0,0,1,1])

model = Logistic_regression()

model.fit(X,y)
model.predict(X)
```
