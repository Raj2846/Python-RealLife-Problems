### Creating the GD with use of python for Linear Regression

```python
import numpy as np
```

```python
class LinearRegression:
    def __init__(self,learning_rate=0.01,iterat=1000):
        self.y_int = None  # b
        self.coeff = None  # m
        self.learning_rate = learning_rate  # alpha used in formula to calcuation
        self.iterat = iterat  # how many time to repeat the process to get the minima

    def fit(self, X_train, y_train):
        r, c = X_train.shape

        # step 1
        self.y_int = 0
        # as there can me multiple intput columns we dont direct use 0 for coeff and instead use array of 0, c represent no.of columns [0,0,0,....0]
        self.coeff = np.zeros(c)

        # we repeat this whole process -> Gradient Descent
        for i in range(0,self.iterat):
            # step 2
            y_pred = self.y_int * np.dot(X_train, self.coeff)

            # step 3 calculate derivaties
            d_yInt = (1/r) * np.sum(y_pred - y_train)
            d_coeff = (1/r) * np.dot(X_train.T,(y_pred - y_train))
            
            # step 4 conversion theorym
            self.y_int = self.y_int - self.learning_rate * d_yInt
            self.coeff = self.coeff - self.learning_rate * d_coeff
        

    def predict(self, x_test):
        y_predict = self.y_int + np.dot(x_test,self.coeff)
        return y_predict
```

```python
import pandas as pd
```

```python
X= np.array([[2],[3],[4],[5]])
y= np.array([4,6,8,10])

model = LinearRegression()
model.fit(X,y)
model.predict(X)
```
