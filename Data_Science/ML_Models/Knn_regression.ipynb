```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("diamonds.csv")
df.head()
```

```python
df.info()
```

```python
df.drop(columns=['Unnamed: 0'],inplace=True)
```

```python
df['cut'].unique()
```

```python
df['color'].unique()
```

```python
df['clarity'].unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
# categories contain the priority wise format of the values into the columns we want to encode 
ord = OrdinalEncoder(categories=[['Fair','Good','Very Good','Premium','Ideal'],['D','E','F','G','H','I','J'],['I1','SI2','SI1','VS2','VS1','VVS2','VVS1','IF']])
```

```python
# here we define the column where the encoding will be applied after calculating the value in the categories in the upper cell
ord.fit_transform(df[['cut','color','clarity']])
```

```python
# then changed the values in the main df with the encoded values 
df[['cut','color','clarity']] = ord.fit_transform(df[['cut','color','clarity']])
```

```python
df.head()
```

```python
# now the encoding is done we will do feature scaling
```

```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler
```

```python
mn = MinMaxScaler()
mn.fit_transform(df.iloc[:,1:]).shape
```

```python
scale_cols = ["carat", "cut", "color", "clarity", "depth", "table", "x", "y", "z"]
df[scale_cols] = mn.fit_transform(df[scale_cols])
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df[scale_cols]
Y = df.iloc[:,6]
```

```python
X_train , X_test , y_train , y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
from sklearn.neighbors import KNeighborsRegressor
```

```python
knn = KNeighborsRegressor()
```

```python
knn.fit(X_train,y_train)
```

```python
y_pred = knn.predict(X_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test,y_pred)
```
