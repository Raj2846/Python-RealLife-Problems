## Complete Case Analysis

Complete case analysis also called "list-wise deetion" of cases , consists in discarding observation where values in any of the variable are missing
<br>
CCA means literally analyzing onlt those observation for which there is information i all of the variable in the dataset

<br>
observation  = rows
variable  =  columns 
<br><br>
Assumption for CCA
<br>
1. Missing completely at random  : then remove that data if completely randomly then only 


## Advantage
<br>
1.Easy to implement as no data manipulation required<br>
2.Prevent variable distributions

## Disadvantage
1.It can exclude a large fraction of the original dataset<br>
2.Excluded observation could be informative for the analysis <br>
3.When using our models in production , the model will not know how to handle missing data

## When to use CCA ?
1. missing completely at random<br>
2. Less than 5% missing data<br>

#### Mean and Median Imputation
<br>
If the distrubation is normal than use mean imputation <br>
If not then median

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df_t = pd.read_csv('titanic_toy.csv')
df_t.head()
```

```python
df_t.isnull().mean()
```

```python
X= df_t.drop(columns=['Survived'])
Y=df_t['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train.shape,X_test.shape
```

```python
X_train.isnull().mean()
```

```python
mean_age = X_train['Age'].mean()
median_age = X_train['Age'].median()

mean_fare = X_train['Fare'].mean()
median_fare = X_train['Fare'].median()
```

```python
X_train['Age_median'] = X_train['Age'].fillna(median_age)
X_train['Age_mean'] = X_train['Age'].fillna(mean_age)

X_train['Fare_median'] =  X_train['Fare'].fillna(median_fare)
X_train['Fare_mean'] =  X_train['Fare'].fillna(mean_fare)
```
