```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("titanic.csv")
df.head()
```

```python
df['number'].unique()
```

```python
df['number_numeric']=pd.to_numeric(df['number'],errors='coerce',downcast='integer')
```

```python
df['number_categorical'] = np.where(df['number_numeric'].isnull(),df['number'],np.nan)

df.head()
```

```python
df['Cabin'].unique()
```

```python
df['Cabin_num'] = df['Cabin'].str.extract('(\d+)')
df['cabin_cat'] = df['Cabin'].str[0]

df.head()
```

```python
df['Ticket_num'] = df['Ticket'].apply(lambda s : s.split()[-1])
df['Ticket_num'] = pd.to_numeric(df['Ticket_num'],errors='coerce',downcast='integer')


df['Ticket_cat'] = df['Ticket'].apply(lambda s : s.split()[-1])
df['Ticket_cat'] = np.where(df['Ticket_cat'].str.isdigit(),np.nan,df['Ticket_cat'])

df.head(10)
```

```python
date['date'] = pd.to_datetime(date['date'])
date.info()
```

```python
# extract year
date['year'] = date['date'].dt.year
date.head(2)
```

```python
# extract month 
date['month'] = date['date'].dt.month
# if want month name then  date['month'] = date['date'].dt.month_name()
date.head
```

```python
# extract days
date['days'] = date['date'].dt.day
date.head(2)
```

```python
# day of the week number
date['day_of_week'] = date['date'].dt.day_of_week
date.head(2)
```

```python
# day of the week name
date['day_of_week'] = date['date'].dt.day_name()
date.head(2)
```

```python
# exttract week of the year
date['date_week'] = date['date'].dt.weekday
date.head(2)
```

```python
# extract time gap between two dates
import datetime

today = datetime.datetime.today()
today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```
