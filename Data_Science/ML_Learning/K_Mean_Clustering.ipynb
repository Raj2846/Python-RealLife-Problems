```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("student_clustering.csv")
df.head()
```

```python
plt.scatter(df['cgpa'],df['iq'])
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []

for i in range(1,11):
    km = KMeans(n_clusters=i)
    km.fit_predict(df)
    wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
```

```python
X= df.iloc[:,:].values
km = KMeans(n_clusters=4)
y_mean = km.fit_predict(X)
```

```python
y_mean
```

```python
#X[boolean indexing(condition wise indexing),which index vlaue to extract]
X[y_mean == 0,0]
```

```python
plt.scatter(X[y_mean == 0,0],X[y_mean == 0,1],color='red')
plt.scatter(X[y_mean == 1,0],X[y_mean == 1,1],color='blue')
plt.scatter(X[y_mean == 2,0],X[y_mean == 2,1],color='green')
plt.scatter(X[y_mean == 3,0],X[y_mean == 3,1],color='yellow')
```
