### Chained Equations for Missing value (MICE algorithm) Iterative Imputer
MICE stands for Multivariate Imputation by Chained Equations

We use it in certain assumption only <br>
Three Categories of missing data <br>
1. Missing Compelete at random (Ceratin rows data are not collected by any reason)
2. Missing at random (here the data is not filled by choice or anything)
3. Missing not at random (Data is removed knowingly)


We use MICE when the data is Missing at random,we get better result on this type of data set

1. Its Accurate (A)
2. Its Slow (D) 
3. Deploy training set on server

#### How it works 
Stage 1<br>
Its only applyes to input columne<br>
1. Fill all the NaN values with the mean of the respective columns
2. Remove all columns missing values (we move from left to right then the value which we filled with the mean value we replace it with NaN again, because we are going to predict that values, we replace from left to right columns flow shall be maintain and no other values must be changed)
<br>
so we consider the remaining rows of other columns as an training data and the rows on the same columns are consider as output columns<br>
then we will select the second column form left to right side with missing values and do the same process again until all the missing data is filled 
<br> This all is stage 1

Stage 2<br>
We will repeat the stage 1 process again and again <br>
so after iteration 0 the data will consist of missing values filled with mean values<br><br>
After iteration 1 the data will consist of the values we got from stage 1 process <br><br>
After that we will minus iteration 0 value with iteration 1 and calculate the difference<br><br>
mose of the values will be 0 but the missing row/colums values will me something <br><br>
We will make other iteration and continue this process of finding the difference until all the values are 0 or almost 0<br><br>
for the 2nd iteration we will take iteration 1 as an base and make the missing value calculated as NaN and the process the stage 1 and get iteration 2 and then find the difference in iteration 1 and iteration 2 <br><br>

Why we consider 0 ot close to 0 , so that we know that we are close to the actual value ,the values is perfect for that column as ther is no difference after all the iteration after calculating mean and stage 1 process
