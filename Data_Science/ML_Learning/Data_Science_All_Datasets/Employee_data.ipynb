```python
df[['Monthly_Salary','Performance_Score']] = df[['Monthly_Salary','Performance_Score']].fillna(df[['Monthly_Salary','Performance_Score']].mean())
```

```python
df
```

```python
df['Department'] = df['Department'].str.strip().str.title()
```

```python
df
```

```python
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'],format='mixed')
```

```python
df
```

```python
df['Tenure_Category'] = df['Experience_Years'].apply(lambda x : 'Fresher' if (x>0 and x<=2) else ('Mid-level' if x > 3 and x < 5 else "Experienced") )
```

```python
df
```

```python
Q1 = np.quantile(df['Monthly_Salary'],0.25)
Q3 = np.quantile(df['Monthly_Salary'],0.75)
IQR = Q3 - Q1
IQR
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
outliers = df[
    (df['Monthly_Salary'] < lower_bound) |
    (df['Monthly_Salary'] > upper_bound)
]
outliers
```
