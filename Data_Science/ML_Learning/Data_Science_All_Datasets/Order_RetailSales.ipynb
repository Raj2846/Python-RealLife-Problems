```python
df_ord['Quantity'] = df_ord['Quantity'].fillna(
    df_ord['Quantity'].mean())
df_ord['Quantity']
```

```python
df_ord.drop_duplicates(ignore_index=True,inplace=True)
```

```python
df_ord
```

```python
df_ord['Date'] = pd.to_datetime(df_ord['Order_Date'],errors='coerce',format='mixed')
```

```python
df_ord['Order_month'] = df_ord['Date'].dt.month
```

```python
df_ord['Order_Year'] = df_ord['Date'].dt.year
```

```python
df_ord.drop(columns=['Date'],inplace=True)
```

```python
df_ord
```

```python
df_ord['Region'] = df_ord['Region'].str.strip().str.title()
```

```python
df_ord
```

```python
merged_df = pd.merge(df_ord,df_pro,how='inner',on='Product_ID')
merged_df
```

```python
merged_df['Sales'] = merged_df['Quantity'] * merged_df['Unit_Price']
```

```python
merged_df
```

```python
import sqlite3

conn = sqlite3.connect("Retail_Sales.db")
```

```python
merged_df.to_sql("Sales_Data",conn,if_exists='replace',index=False)
```

```python
query = "Select * from Sales_Data"

result = conn.execute(query)
x = result.fetchall()
for i in x:
    print(x[:1],sep="\n")
```

```python
query = "Select Order_ID,order_Date,Product_Name,Region,Quantity,Sales from Sales_Data"

result = conn.execute(query)

x = result.fetchall()
for i in x:
    print(x[:1],sep="\n")
```

```python
query = """
Select * from Sales_Data
Where Region = 'West' and Sales > 500 Order by Sales desc
"""

result = conn.execute(query)

x = result.fetchall()
for i in x:
    print(x[:1],sep="\n")
```

```python
query = """
Select sum(Sales) as Total_Sales,sum(Quantity) as Total_Quantity,count(Order_ID) from Sales_Data Group by Region
"""
result = conn.execute(query)

x = result.fetchall()

for i in range(len(x)):
    print(x[i],sep="\n")
```

```python
# Create Orders table
df_ord.to_sql("Orders", conn, if_exists="replace", index=False)

# Create Products table
df_pro.to_sql("Products", conn, if_exists="replace", index=False)
```

```python
query = """
Select * from Orders
"""
result = conn.execute(query)

x = result.fetchall()

for i in range(len(x)):
    print(x[i],sep="\n")
```

```python
query = """
Select * from Products
"""
result = conn.execute(query)

x = result.fetchall()

for i in range(len(x)):
    print(x[i],sep="\n")
```

```python
query = """
Select 
    o.Order_ID,
    p.Product_ID,
    p.Product_Name,
    p.Category,
    o.Region,
    o.Quantity,
    o.Unit_Price from Orders o
    Join Products p ON o.Product_ID = p.Product_ID
"""
result = conn.execute(query)

x = result.fetchall()

for i in range(len(x)):
    print(x[i],sep="\n")
```

```python
query = """
SELECT
    Region,
    SUM(Sales) AS Total_Sales,
    COUNT(Order_ID) AS Total_Orders,
    AVG(Sales) AS Average_Sales
FROM Sales_Data
GROUP BY Region
ORDER BY Total_Sales DESC;
"""

sales_summary = pd.read_sql(query,conn)
sales_summary
```
