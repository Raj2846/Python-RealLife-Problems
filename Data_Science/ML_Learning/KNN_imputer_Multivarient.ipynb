### KNN imputer Working 

Here the imputer calculate the Kth nearest neighour of the missing value form the column(feature) and fill the values with that values <br>
Main is the value of the K means how many neighbor to take into consideration while calculating the value for the missing row <br>
value of the neighbor and divide by the number of neighbor<br>

total of all the value of k / number of values

#### steps to follow

1. Find the Kth nearest neighbor (by finding the distance eudidean distance for this case)
2.  Find the values (har neighbor se uska value le ke mean nikal lena )

3. Distance formula sqrt (x2 - x2)^2 + (y2 - y1)^2 , for 2d coordinate system

We can have multiple missing values so we use other formula that is <br>
nan_eudidean distance = This handles the missing values in the data and calculate the distance<br>
Formula = dist(x,y) = sqrt(weight * sq .distance from present coordinate) , where weight = total No.of coordinates / No.of present coordinate (where all the values are present in both row in same column)

### Advantage(A) & Disadvantage(D)<br>
1. Its more accurate (A)
2. More number of calculation (D)
3. When deploy on server , we will have to upload the whole training set on serve (D)

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.impute import KNNImputer,SimpleImputer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('train.csv')[['Age','Pclass','Fare','Survived']]
df.head(2)
```

```python
df.isnull().mean()*100
```

```python
X = df.drop(columns=['Survived'])
Y= df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train.head()
```

```python
knn = KNNImputer()
#'distance','uniform' difference in uniform we add the values of the kth nearest neighbur and divide it by no.of neighbor (total of kth values)/no.of kth term

# but in distance = we will do the reciprocal of the values so
# (1/(distance of the value calculated) * value + 1/(other disatnce)* value) / 2
```

```python
X_train_trf = knn.fit_transform(X_train)
X_test_trf = knn.transform(X_test)
```

```python
pd.DataFrame(X_train_trf,columns=X_train.columns)
```

```python
lr = LogisticRegression()

lr.fit(X_train_trf,y_train)

y_pred = lr.predict(X_test_trf)

accuracy_score(y_test,y_pred)
```

```python
# doing the same thing with mean imputation using simple imputation class

si =SimpleImputer()

X_train_trf2= si.fit_transform(X_train)
X_test_trf2 = si.transform(X_test)
```

```python
lr = LogisticRegression()

lr.fit(X_train_trf2,y_train)

y_pred2 = lr.predict(X_test_trf2)

accuracy_score(y_test,y_pred2)
```
