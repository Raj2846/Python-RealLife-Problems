Decision Tree 

Its an ladder of nested if else statement <br>
 

Steps of Decision Tree

1. Begin with your training dataset , which should have some feature variables and classification or regression output
2. Determin the "best feature" in the dataset to split the data on (selecting best feature we will learn later)
3. Split the data into subset that contain the correct values for this best feature . This splitting basically defines a node on the tree i.e each node is a splitting point based on certain feature from our data.
4. Recursively generate new tree node by using the subset of data created from step 3

### Decision Tree : Entropy
<br>
Entropy is the measure of purity/inpurity or we can say measure of disorder<br><br>
More Knowledge Less entropy : The more you get knowledge from the dataset the less entropy it has and vice/versa

Mathematical Formula for entropy : E(S) = sigma of - pi log2 pi <br><br>
Where pi is simply the frequentist probability of an element/class 'i' in our data<br><br>

Example : if our data has only 2 class labesls YES and NO <br>
E(D) = -pYES log2(pYES) -pNO log2(pNO)   <br><br>
The more class we have the number of pi increases according to class

Observation :
1. More the uncertainty more is entropy
2. For 2 class problems the min entropy is 0 and max is 1
3. For more than 2 class min is 0 max is greater then 1
4. Both log2 or loge can be used to calculate entropy

Entropy for numeric or continuous variables : simple trick , which ever dataset has less peak when ploted in graph has more entropy

### Decision Tree :Information Gain

1. Information gain is a metric used to train Decision tree specifically , this metric measures the quality of split
2. The Information gain is based on the decrease in the entropy after a dataset is split on an attribute
3. Constructing a decision tree is all about finding sttribute that returns the highest information gain 

Formula :
1. Informatin gain  = E(Parent) - {Weignted Average} * E(children)
E = Entropy<br>
parent = Complete dataset before split <br>
Children = divide the dataset based on any colums

Supposer you have an dataset of 14 rows and 4 input and 1 output columns <br>
Input columns = [Outlook,temperature,Humidity,Windy]<br>
Output = [play tennis] its in yes or no 2 class classification problem 
<br>
The outlook columns has 3 types of values [Sunny,Overcast,Rain]
<br>

1. Step 1 <br>
So according to formula we will find E(Parent) i.e whole dataset <br>
E(p) = -pYes log2(pYES) - pNO log2(pNO)
here suppose E(p) = 0.94

2. Step 2<br>
So the whole dataset will be splited on base of the outlook colums i.e into 3 partes 1. Sunny 2. Overcast  3. Rain <br>
each partes contain the all the columns data this 3 parts are considers as child dataset
<br><br>
Here we will calculate the entropy of the child dataset so after calculating we will have 3 child entopy, one for sunny dataset , one for overcast ,one for rain<br>
Entropy Sunny = 0.97 (Conatin 5 rows so 5/14 in formula 2 yes 3 no)<br>
Entropy Overcast = 0 (Conatin 5 rows so 4/14 in formula all yes)<br>
Entropy rain = 0.97 (Conatin 5 rows so 5/14 in formula 3 yes 2 no)


3. Step 3<br>
Here we will calculate Weighted Entropy of children i.e sunny ,overcat,rain<br>
Weighted Entropy = 5/14*0.97 + 4/14 * 0 + 5/14 * 0.97 <br>
W.E(Children) = 0.69

4. Step 4 <br>
Calculate the Informatino gain <br>

IG = E(parents) - {weighted average} * E(children)<br>
IG = 0.97 - 0.69 = 0.28 <br><br>

So the Information gain (the decrease in entropy/impurity) when spliting the data on the basis of Outlook columns is 0.28

5. Step 5<br>
We will calculate the infotmation gain for all the columns. Which ever columns has the maximum decrease in entropy , the algorithm will select that columns to split the data 

6. Step 6<br>
Find Information Gain Recursively <br>
DEcision tree then applies a recursive greedy search algorithm in top bottom fashion to find informatino gain ata every level of the tree 

<br><br>
Once a leaf node is reached (entropy = 0) no more splitting is done

#### Handling numberic data For Decision Tree <br>
Here the input columns is numeric 

1. Sort the data on the bases of numeric columns 
2. Split the entire data on the basis of each numeric value into 2 dataset
3. Find out Entropy for each of this child dataset 
4. Find weighted entropy as we are getting 2 dataset for each numeric values scenerio
5. Find Max entropy from all the scenerio and set it as an spliting criteria 
6. Do this recursively until we get to leaf values
