```python
import pandas as pd
import numpy as np
import seaborn as sns

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('train.csv')[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
X = df.iloc[:,0:4]
Y = df.iloc[:,-1]
```

```python
np.mean(cross_val_score(LogisticRegression(),X,Y,scoring='accuracy',cv=20))
```

### Applying Feature Construction

```python
X['Family_size'] = X['SibSp'] + X['Parch'] + 1
```

```python
X.head()
```

```python
def fun(num):
    if num == 1:
        #alone
        return 0
    elif num > 1 and  num <= 4:
        #small family
        return 1
    else:
        #large family
        return 2
```

```python
X['family_type'] = X['Family_size'].apply(fun)
```

```python
X.sample(5)
```

```python
X.drop(columns=['SibSp','Parch','Family_size'],inplace=True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,Y,scoring='accuracy',cv=20))
```

After doing feature Construction we can see that there is little change in the accuracy on same model

## Feature Splitting

```python
df_s['Title'] = df_s['Name'].str.split(',',expand=True)[1].str.split('.',expand=True)[0]
```
