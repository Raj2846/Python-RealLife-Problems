Basic requirements : Decision Tree

Here we train multiple group of decision tree therefore called Forest , Because its use BAGGING (Bootstrapped Aggregation : Random Data sampling)techinique its called Random. In BAGGING we can use any base algorithm but if we use Decision Tree as base algorithm then its called Random Forest.

Bagging : Consider we have an dataset of 1000 rows so the step 1 : we pick any base model like k-mean, svm, decision tree , so for this is random forest we use decision tree as base model  
1. Step 1 : Booth Strapping<br>
So there are multiple decision tree like DT1, DT2, DT3.......DT100 <br>
now of this dataset we do Sampling (Subset of dataset): there are 3 types of sampling :
<br>
1.Row sampling : we decide to give every base model 500 rows only , now how to decide which 500 rows to give :-Its randomly selected , no This sample dataset is given to Base Model DT1 <br>
Here there are two type of ROW sampling : With replacement(Here the row can be selected again after its selected once , so chances of duplication ) , without replacement(there once row is selected its not used again)
<br>
2.Column Sampling : Here we decide to give only 5 random columns from 10 columns to the 1st base model and continue for others models <br>
3.Combo of Row and Columns sampling.

2. Step 2 : Aggregation<br>
After Sampling all the models are trained on that datasets. <br>
Now for Prediction we Use Aggregation : Now a new query point came and that is given to all the decision tree i.e 100 , now consider its Classification problem , so we check how many Decision Tree has predicted Output as "0" and "1".<br>
After that majority is used to select the which output to consider , so if 57 model predicts as 0 then we consider final predicts as 0 
<br>
If its Regression Problem then : Here instead of 0 or 1 we will get predict as an number so , we will calculate the mean of all the base models predicted output and consider it as our final output i.e (mean of all models)

Bias Variance : Bias is called when model does not give proper result on training data .bias is high there, variance is called training data is muguped(not learning pattern) so when new test data is given the result is not proper its called high variance<br>
So we need low bias and low variance , but they are inverse to each other so high bias low variance and vice/versa<br>

Low bias high variance : Full grown DT,SVM,KNN<br>
High bias low variance : LR<br>

Random Forest when given an LBHV(Low bias high variance) then it converts it into LBLV(low bias low variance) algortihm <br>
How does it happen : In random forest we create subset of the dataset and train the base models , so there is always low bias as for variance :since we are using random forest there are multiple models in it so the impact of the outliers/noisey point is distributed into the models so the impact is low , so the variance is reduced , so the output is LBLV

In Bagging the sampling are done according to row or columns sampling , its pre decided on which columns the sampling will happen , so its tree level sampling. <br>
In Random Forest node level splitting is done so on bases of root node the furthure sampling is done randomnes is more here as compare to Bagging
