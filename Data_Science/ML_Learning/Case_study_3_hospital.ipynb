## Part B - Chapter 5: Data Type Conversion

#### Remove leading/trailing spaces from Patient_Name.

```python
df["Patient_Name"] = df["Patient_Name"].str.strip()
df.head()
```

#### Convert Age safely to a nullable integer. Invalid age text must become missing.

```python
df["Age"] = pd.to_numeric(df["Age"],errors='coerce').astype("Int32")
```

```python
df.head()
```

####  Clean Fee values containing commas/spaces and convert Fee to numeric.

```python
df["Fee"] = pd.to_numeric(df["Fee"].astype(str).str.replace(',','').str.strip(),errors='coerce').astype("Int32")
df["Fee"].dtypes
```
